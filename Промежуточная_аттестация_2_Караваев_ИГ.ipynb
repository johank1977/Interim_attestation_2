Чистим данные, преобразуем типы данных соответственно в текст, дата, int или float, заменяем пропуски и NaN на 0

```python
import pandas as pd
import numpy as np

def clean_dataframe(df):
    """
    Чистит DataFrame: заменяет NaN/NaT на 0, преобразует строки с числами в числовые типы,
    оставляя даты без изменений, а текстовые данные приводит к str.
    """
    # Копируем DataFrame, чтобы не менять оригинал
    cleaned_df = df.copy()

    # Заменяем все NaN и NaT на 0
    cleaned_df.fillna(0, inplace=True)


    for column in cleaned_df.columns:
        # Пропускаем столбцы с датами
        if pd.api.types.is_datetime64_any_dtype(cleaned_df[column]):
            continue

        # Пропускаем уже числовые столбцы
        if pd.api.types.is_numeric_dtype(cleaned_df[column]):
            # Для числовых столбцов явно приводим к int/float при необходимости
            if cleaned_df[column].dtype == 'float64' and (cleaned_df[column] % 1 == 0).all():
                cleaned_df[column] = cleaned_df[column].astype(int)
            continue

        # Пытаемся преобразовать в числовой тип
        converted_series = pd.to_numeric(cleaned_df[column], errors='coerce')

        # Если после преобразования есть не-NaN значения, значит столбец содержит числа
        if not converted_series.isna().all():
            # Заменяем NaN, появившиеся из нечисловых строк, на 0
            converted_series.fillna(0, inplace=True)
            # Оптимизируем тип: если все числа целые, преобразуем float в int
            if (converted_series % 1 == 0).all():
                converted_series = converted_series.astype(int)
            cleaned_df[column] = converted_series
        else:
            # Если преобразование не удалось — приводим к строковому типу
            #cleaned_df[column] = cleaned_df[column].astype(str)

            cleaned_df['region'] = cleaned_df['region'].astype('string')
            cleaned_df['exemption_criteria'] = cleaned_df['exemption_criteria'].astype('string')


    return cleaned_df


df = pd.DataFrame(df_reduced)

print("До очистки:")
print(df)
print("\nТипы данных до:")
print(df.dtypes)

cleaned_df = clean_dataframe(df_reduced)

print("\nПосле очистки:")
cleaned_df.head()
print("\nТипы данных после:")
print(cleaned_df.dtypes)
```

2. Постройте гистограммы для трех колонок, которые, на ваш взгляд, могут содержать полезную информацию. Выясните, есть ли в этих колонках выбросы и аномалии (значения, которые сильно отличаются от остальных значений или кажутся странными/некорректными).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


# Создаём пример DataFrame (замените на ваши данные)
np.random.seed(42)
df = pd.DataFrame({
    'age': np.random.normal(35, 12, 1000),  # возраст
    'salary': np.random.lognormal(10, 0.8, 1000),  # зарплата (асимметричное распределение)
    'experience_years': np.random.exponential(5, 1000)  # опыт работы
})

# Добавляем несколько выбросов для демонстрации
df.loc[10] = [85, 500000, 40]
df.loc[20] = [20, 200000, 35]
df.loc[30] = [15, 80000, 50]

# Настройка визуализации
plt.figure(figsize=(15, 12))

# Гистограмма 1: Возраст
plt.subplot(3, 2, 1)
df['age'].hist(bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Распределение возраста')
plt.xlabel('Возраст')
plt.ylabel('Количество наблюдений')

# Boxplot для возраста
plt.subplot(3, 2, 2)
sns.boxplot(y=df['age'], color='lightblue')
plt.title('Boxplot: Возраст')
plt.ylabel('Возраст')

# Гистограмма 2: Зарплата
plt.subplot(3, 2, 3)
df['salary'].hist(bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
plt.title('Распределение зарплаты')
plt.xlabel('Зарплата')
plt.ylabel('Количество наблюдений')

# Boxplot для зарплаты
plt.subplot(3, 2, 4)
sns.boxplot(y=df['salary'], color='lightgreen')
plt.title('Boxplot: Зарплата')
plt.ylabel('Зарплата')

# Гистограмма 3: Опыт работы
plt.subplot(3, 2, 5)
df['experience_years'].hist(bins=30, color='salmon', edgecolor='black', alpha=0.7)
plt.title('Распределение опыта работы')
plt.xlabel('Опыт (лет)')
plt.ylabel('Количество наблюдений')

# Boxplot для опыта работы
plt.subplot(3, 2, 6)
sns.boxplot(y=df['experience_years'], color='salmon')
plt.title('Boxplot: Опыт работы')
plt.ylabel('Опыт (лет)')

plt.tight_layout()
plt.show()
```
