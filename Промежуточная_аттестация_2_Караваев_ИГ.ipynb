Преобразуем формат данных для удобстав работы и полученный результат запишем в датафрейм *cleaned_df* с которым и будем дальше работать.

```python
import pandas as pd
import numpy as np

def clean_dataframe(df):
    """
    Чистит DataFrame: заменяет NaN/NaT на 0, преобразует строки с числами в числовые типы,
    оставляя даты без изменений, а текстовые данные приводит к str.
    """
    # Копируем DataFrame, чтобы не менять оригинал
    cleaned_df = df.copy()


    cleaned_df['region'] = cleaned_df['region'].astype('string')
    cleaned_df['exemption_criteria'] = cleaned_df['exemption_criteria'].astype('string')
    cleaned_df['okved_section'] = cleaned_df['okved_section'].astype('string')

    # Заменяем все NaN и NaT на 0
    cleaned_df.fillna(0, inplace=True)


    for column in cleaned_df.columns:
        # Пропускаем столбцы с датами
        if pd.api.types.is_datetime64_any_dtype(cleaned_df[column]):
            continue

        # Пропускаем уже числовые столбцы
        if pd.api.types.is_numeric_dtype(cleaned_df[column]):
            # Для числовых столбцов явно приводим к int/float при необходимости
            if cleaned_df[column].dtype == 'float64' and (cleaned_df[column] % 1 == 0).all():
                cleaned_df[column] = cleaned_df[column].astype(int)
            continue

        # Пытаемся преобразовать в числовой тип
        converted_series = pd.to_numeric(cleaned_df[column], errors='coerce')

        # Если после преобразования есть не-NaN значения, значит столбец содержит числа
        if not converted_series.isna().all():
            # Заменяем NaN, появившиеся из нечисловых строк, на 0
            converted_series.fillna(0, inplace=True)
            # Оптимизируем тип: если все числа целые, преобразуем float в int
            if (converted_series % 1 == 0).all():
                converted_series = converted_series.astype(int)
            cleaned_df[column] = converted_series
        else:
            # Если преобразование не удалось — приводим к строковому типу
            cleaned_df[column] = cleaned_df[column].astype('string')


    return cleaned_df


df = pd.DataFrame(df_cleaned)

print("До очистки:")
print(df)
print("\nТипы данных до:")
print(df.dtypes)

cleaned_df = clean_dataframe(df_cleaned)

print("\nПосле очистки:")
cleaned_df.head()
print("\nТипы данных после:")
print(cleaned_df.dtypes)
```

В колонке *okved_section* символьное обозначение секции заменим на числовое (порядковый номер буквы в алфавите), сделаем это в новую колонку, после чего удалим текстовую.

```python
def letter_to_number(letter):
    if isinstance(letter, str) and len(letter) == 1:
        if 'a' <= letter <= 'z':
            return ord(letter) - ord('a') + 1
        elif 'A' <= letter <= 'Z':
            return ord(letter) - ord('A') + 1
    return None  # для не‑букв или некорректных значений

cleaned_df['okved_section_int'] = cleaned_df['okved_section'].apply(letter_to_number).astype('Int64')
cleaned_df.head()
```

Вычислим и визуализируем матрицу корреляции между всеми 20 колонками. Найдем между какими колонками имеется высокая корреляция.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Оставляем только числовые колонки для корреляционного анализа
numeric_df = cleaned_df.select_dtypes(include=[np.number])


# Вычисляем матрицу корреляции (по умолчанию — корреляция Пирсона)
correlation_matrix = numeric_df.corr()

print("Размер матрицы корреляции:", correlation_matrix.shape)
print("\nМатрица корреляции (первые 5×5):")
print(correlation_matrix.iloc[:5, :5].round(3))
```

 Визуализация матрицы корреляции

```python
plt.figure(figsize=(14, 12))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    square=True,
    cbar_kws={"shrink": .8},
    linewidths=.5
)
plt.title('Матрица корреляции между 20 колонками DataFrame', fontsize=16, pad=20)
plt.tight_layout()
plt.show()
```

Поиск пар с высокой и низкой корреляцией

```python
# Находим пары с |корреляция| > 0.8
high_corr_pairs = []
threshold = 0.8

for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]
        if abs(corr_value) >= threshold:
            high_corr_pairs.append((
                correlation_matrix.columns[i],
                correlation_matrix.columns[j],
                corr_value
            ))

print(f"Пары с высокой корреляцией (|r| ≥ {threshold}):")
if high_corr_pairs:
    for col1, col2, corr in high_corr_pairs:
        print(f"{col1} и {col2}: r = {corr:.3f}")
else:
    print("Пары с корреляцией |r| ≥ 0.8 не найдены.")

# Находим пары с отрицательной корреляцией
negative_corr_pairs = []
threshold = -0.5  # порог: ищем корреляцию ≤ -0.5

for i in range(len(correlation_matrix.columns)):
    for j in range(i + 1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]
        if corr_value <= threshold:
            negative_corr_pairs.append((
                correlation_matrix.columns[i],
                correlation_matrix.columns[j],
                corr_value
            ))

# Сортируем по возрастанию (самая сильная отрицательная — первая)
negative_corr_pairs_sorted = sorted(negative_corr_pairs, key=lambda x: x[2])

print(f"Пары с отрицательной корреляцией (r ≤ {threshold}):")
if negative_corr_pairs_sorted:
    for col1, col2, corr in negative_corr_pairs_sorted:
        print(f"{col1} и {col2}: r = {corr:.3f}")
else:
    print("Пары с корреляцией r ≤ -0.5 не найдены.")
```

*   Обнаружена сильная положительная корреляция между *inn* и *region_taxcode* (r = 0.963), но эта зависимость логична, поскольку ИНН и регион уплаты налогов практически одно и то же, за некторым исключением для компаний компаний, где регион ИНН (первые 2 цифры), может отличаться от кода региона (например, для ИНН 99,50, 77 - это Москва, а регион только 77);
*   Также обнаружена положительная корреляция между *filed* и *articulated* (r = 0.871), объяснить это можно только тем, что колонки содержат бинарные данные.
*   Также обнаружена положительная корреляция между *filed* и *totals_adjustment* (r = 0.832), объяснить это можно только тем, что колонки содержат бинарные данные.

Пары с отрицательной корреляцией r ≤ -0.5 не найдены.



Визуализация коррелирующих пар (диаграммы рассеивания)

```python
# Сортируем пары по убыванию модуля корреляции
high_corr_pairs_sorted = sorted(high_corr_pairs, key=lambda x: abs(x[2]), reverse=True)

# Берём топ‑3 пары
top_pairs = high_corr_pairs_sorted[:3]

# Строим графики
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
axes = axes.ravel()

for idx, (col1, col2, corr) in enumerate(top_pairs):
    sns.scatterplot(
        data=numeric_df,
        x=col1,
        y=col2,
        alpha=0.6,
        ax=axes[idx]
    )
    axes[idx].set_title(f'{col1} vs {col2}\nr = {corr:.3f}')
    axes[idx].set_xlabel(col1)
    axes[idx].set_ylabel(col2)
    axes[idx].grid(True, alpha=0.3)

plt.suptitle('Диаграммы рассеивания для пар с высокой корреляцией', fontsize=16)
plt.tight_layout()
plt.show()
```

Как было выяснено ранее, выводы из поиска пар с высокой корреляцией визуализированы и  выводы подтверждаются.
